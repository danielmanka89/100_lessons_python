# How to Create a Variable

> Use the = sign to put a value into a variable.

- name = "Daniel"
- age = 25
- city = "Chicago"

```python
name = 'Daniel'
print(name)
```

```python
name = "Daniel"
age = 36
city = 'Chicago'

print(name)
print(age)
print(city)
```

```python
movie = 'The Suits'
print(movie)
```

```python
fruit = 'grapes'
quantity = 10
price = 2.99

print(fruit)
print(quantity)
print(price)
```

```python
fruit = 'grapes'
quantity = 10
price = 2.99

print(fruit, quantity, price)
```

```python
x = 10
print(x)
x = 20
print(x)
x = 30
print(x)
```
