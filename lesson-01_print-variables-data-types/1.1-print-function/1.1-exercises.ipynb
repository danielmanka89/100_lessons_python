# STEP-04

## The sep Parameter (Proper Teaching)

### What You Already Know

> You know that commas , add spaces between items:
- print("apple", "banana", "cherry")
- Output: apple banana cherry

### What if you DON'T want spaces?

> What if you want:

- apple-banana-cherry
- apple|banana|cherry
- apple,banana,cherry
- applebananacherry (no space, no symbol)

### The Answer: Use sep

> sep stands for separator.
- It controls what goes BETWEEN your items.

### Changing sep

> You can change the separator to anything you want:

> Dash:
- print("apple", "banana", "cherry", sep="-")
- Output: apple-banana-cherry

> Pipe:
- print("apple", "banana", "cherry", sep="|")
- Output: apple|banana|cherry

> Comma and space:
- print("apple", "banana", "cherry", sep=", ")
- Output: apple, banana, cherry

> Nothing (no separator)
- print("apple", "banana", "cherry", sep="")
- Output: applebananacherry

### EXERCISE

> Print these three words with a | (pipe symbol) between them instead of spaces:

- apple banana cherry

> EXPECTED OUTPUT
- apple|banana|cherry

```python
print("apple","banana","cherry",sep="|")
```

### PRACTICE EXERCISES

> Write code to print: redbluegreen (no separator)

```python
print("red","blue","green", sep="-")

print("red","blue","green", sep=", ")

print("red","blue","green", sep="")
```

# STEP-05

# The end Parameter

> What you already know?

- You know that print() shows things on screen
- You also know that after each print() python automatically move to a next line

### For Example:

- print("Hello")
- print("World")
> OUTPUT:
 - Hello
 - World

> See how world appears on the line? That's because print() automatically adds a "new line" at the end.

### What is end?

> end is a parameter that controls what comes at the end of print() statement

> By default, end="\n" (the \n means "new line")

> That's why Python goes to the next line after printing.

###  Changing end

> you can change and do anything you want.

```python
# No new line, nothing at the end

print("Hello", end="")
print("World")
```

```python
# Space at then end

print("Hello", end=" ")
print("World")
```

```python
# Exclamation mark at the end

print("Hello", end="!!!")
print("World")
```

```python
# Default behavior

print("Hello", end="\n")
print("World")
```

## The difference between sep and end

> sep

- what goes between items
- print("a","b", sep="-")
- output: a-b

> end
- what goes at the end
- print("a", end="!")
- output: a!

# STEP 7

## Special Characters - Tab Space (\t)

> What is \t?

- \t adds a tab space (like pressing the Tab key on your keyboard).
- A tab space is usually 4 to 8 spaces. It helps align text neatly.

```python
print("Name:\tDaniel")
```

```python
print("Item\tPrice")
print("Apple\t$1")
print("Banana\t$2")
```

```python
print("A\tB\tC")
```

```python
print("Name:\tDaniel\nAge:\t36\nCity:\tChicago")
```
